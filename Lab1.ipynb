**3. Generating a sinusoid**

3.1. Create a sinusoid of the same duration than the reference signal and same frequency than the fundamental frequency. Plot it along with the audio signal of the reference recording. Plot the same time segment that 2.3. Note that the sampling rate should be the same for both signals. Try to find the values of Amplitude and initial Phase that make the two signal segments plotted to match as close as possible (do not attempt to imitate the whole signal). Matching the phase is a bit tricky.



```python
# Write the code here
import numpy as np
A = 0.25155567                      ##It's the highest point in the previous graphic
f0 = audafrequency                  ## It's the theoric frequency that it's very similar to the graph one
phi = 49*np.pi/72                   ##We made some manual calculations and approximate to be the most loyal for the previous graphic
fs = 44100                          ##Our sampling rate
t = np.arange(0, 4, 1.0/fs)         ##We prepare the time to calculate all the x
x = A * np.cos(2*np.pi*f0*t + phi)  ##We apply the function to create the signal
print(x)
plot_signals(x,fs, startpoint, endpoint)
```

3.2. What are the main differences between the two signals.

The main difference between both signals is that the second one is more symmetrical and cleaner than the first one. The reason is that the first one is a real signal extracted from an instrument, meaning that other elements interact with these signals, such as the wind, other frequencies, or the quality of the record. A real signal can't be equal to the cosine wave, as this one is idealized.
